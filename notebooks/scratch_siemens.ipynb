```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

siemens_path = RAW_DATA_DIR / "Beispieldatensatz Siemens S7 SPS.csv"
beckhoff_path = RAW_DATA_DIR / "Beispieldatensatz Beckhoff SPS.csv"
```

```python
# Load the CSV files into DataFrames.
siemens_data = pd.read_csv(siemens_path, sep=";")
beckhoff_data = pd.read_csv(beckhoff_path, sep=";")


print("PLC datasets loaded successfully.")
```

```python
def describe_raw_data(data, dataset_name):
    """Display an overview of a raw PLC dataset."""

    print(f"\n{'=' * 50}")
    print(f"Dataset: {dataset_name}")
    print("=" * 50)

    print("\n1. Shape")
    print(f"Rows: {data.shape[0]:,}")
    print(f"Columns: {data.shape[1]}")

    print("\n2. Column names")
    print(data.columns.tolist())

    print("\n3. Data types")
    print(data.dtypes.to_string())

    print("\n4. Missing values")
    print(data.isna().sum().to_string())

    print("\n5. Duplicate rows")
    print(f"Count: {data.duplicated().sum():,}")

    print("\n6. First five rows")
    print(data.head().to_string())

    print("\n7. Basic statistics")
    print(data.describe(include="all").to_string())
```

```python
describe_raw_data(siemens_data, "Siemens S7")
```

```python
def inspect_raw_data(data):
    """Inspect timestamps, selected columns, and missing values."""

    print("First five timestamps:")
    print(data["timestamp"].head().to_list())

    print("\nTimestamp differences in milliseconds:")
    print(data["timestamp"].diff().describe())

    print("\nExtra column values:")
    print(data["Unnamed: 143"].value_counts(dropna=False))

    print("\nExample object-column values:")
    for column in ["M34.4", "M24.7", "M25.3"]:
        print(f"{column}: {data[column].unique()[:10]}")

    print("\nRows with missing values:")
    missing_rows = data[data.isna().any(axis=1)]
    print(f"Number of affected rows: {len(missing_rows)}")
    print(missing_rows.head(3).to_string())
```

```python
inspect_raw_data(siemens_data)
```

```python

# DATA CLEANING: REMOVE A HIGHLY INCOMPLETE ROW
#
# Why are we removing this row?
# 1. The raw Siemens dataset contains 6,283 rows and 143 columns.
# 2. Row index 6282 contains 118 missing values out of 143 columns.
# 3. Therefore, only 25 values are available in this row, including
#    the timestamp and some PLC signal values.
# 4. Most of the information in this row is unavailable. Keeping it
#    could introduce incomplete observations into later analysis,
#    such as signal processing, feature engineering, and stop detection.
# 5. Since this is only one row, removing it has a limited impact on
#    the dataset size. However, we should still acknowledge that its
#    removal discards the 25 available values as well.
# 6. We preserve the original dataset (siemens_data) and remove the
#    row only from our working copy (siemens_clean).
#
# Decision:
# Remove this row because most of its values are missing.
# This decision is based on the observed missing-value pattern,
# not on an assumption that the machine was stopped.

siemens_clean = siemens_clean.drop(index=6282).copy()

# Verify the result.
print("Original dataset shape:", siemens_data.shape)
print("Working dataset shape:", siemens_clean.shape)
print(
    "Rows containing missing values:",
    siemens_clean.isna().any(axis=1).sum()
)
```

```python

# Inspect object columns to identify their actual values.
object_columns = siemens_clean.select_dtypes(
    include=["object"]
).columns

print("Number of object columns:", len(object_columns))

for column in object_columns:
    values = siemens_clean[column].dropna().unique()

    if all(isinstance(value, (bool,)) for value in values):
        print(f"{column}: Boolean values")
    else:
        print(f"{column}: {values[:10]}")
```

```python

# Identify columns stored as object.
object_columns = siemens_clean.select_dtypes(
    include=["object"]
).columns

# Count missing values in these columns.
missing_object_values = siemens_clean[object_columns].isna().sum().sum()

# Check whether all non-missing values are Boolean.
all_boolean = all(
    siemens_clean[column]
    .dropna()
    .map(lambda value: isinstance(value, bool))
    .all()
    for column in object_columns
)

print("Number of object columns:", len(object_columns))
print("Missing values in object columns:", missing_object_values)
print("All non-missing values are Boolean:", all_boolean)
```

```python

# Convert object columns containing Boolean values to bool dtype.
object_columns = siemens_clean.select_dtypes(
    include=["object"]
).columns

siemens_clean[object_columns] = siemens_clean[object_columns].astype(bool)

# Verify the conversion.
print("Data types after conversion:")
print(siemens_clean.dtypes.value_counts())

print("\nExample converted columns:")
print(siemens_clean[["M34.4", "M24.7", "M25.3"]].dtypes)
```

```python

# Identify columns that contain only one unique value.
constant_columns = [
    column
    for column in siemens_clean.columns
    if siemens_clean[column].nunique() <= 1
]

print("Number of constant columns:", len(constant_columns))
print("\nConstant columns and their values:")

for column in constant_columns:
    print(f"{column}: {siemens_clean[column].iloc[0]}")
```

```python

# Summarize the constant columns.
constant_summary = pd.DataFrame({
    "column": constant_columns,
    "data_type": [
        str(siemens_clean[column].dtype)
        for column in constant_columns
    ],
    "constant_value": [
        siemens_clean[column].iloc[0]
        for column in constant_columns
    ],
    "unique_values": [
        siemens_clean[column].nunique()
        for column in constant_columns
    ],
})

print(constant_summary.to_string(index=False))
```

```python

# Count columns with more than one unique value.
variable_columns = [
    column
    for column in siemens_clean.columns
    if siemens_clean[column].nunique() > 1
]

print("Total columns:", len(siemens_clean.columns))
print("Constant columns:", len(constant_columns))
print("Variable columns:", len(variable_columns))


# ### What this tells us

# - **Constant columns:** No variation in this recording.
# - **Variable columns:** Change at least once during the recording.
# - Variable does not automatically mean useful, and constant does not automatically mean irrelevant.

# After you share these three numbers, we'll inspect the variable columns and then review which constant signals are referenced by the original feature-engineering code. Only after that will we decide what to remove.
```

```python

# Summarize how many distinct values each variable column contains.
variable_summary = pd.DataFrame({
    "column": variable_columns,
    "data_type": [
        str(siemens_clean[column].dtype)
        for column in variable_columns
    ],
    "unique_values": [
        siemens_clean[column].nunique()
        for column in variable_columns
    ],
})

# Show columns with the fewest distinct values first.
variable_summary = variable_summary.sort_values(
    by="unique_values"
)

print(variable_summary.to_string(index=False))
```

```python

# Find the largest gap between consecutive timestamps.
timestamp_intervals = siemens_clean["timestamp"].diff()
largest_gap_index = timestamp_intervals.idxmax()

# Display the observations immediately before and after the gap.
rows_around_gap = siemens_clean.loc[
    [largest_gap_index - 1, largest_gap_index]
]

print("Largest gap (milliseconds):")
print(timestamp_intervals.loc[largest_gap_index])

print("\nAll columns before and after the gap:")
print(rows_around_gap.T.to_string())
```

```python

# Calculate time between consecutive PLC observations.
timestamp_intervals = siemens_clean["timestamp"].diff()

# Exclude the first row because it has no previous timestamp.
timestamp_intervals = timestamp_intervals.dropna()

print("Median interval (ms):", timestamp_intervals.median())
print("Maximum interval (ms):", timestamp_intervals.max())

for threshold in [100, 500, 1000]:
    count = (timestamp_intervals > threshold).sum()

    print(
        f"Intervals greater than {threshold} ms: {count}"
    )
```

```python

# Calculate the time between consecutive observations.
timestamp_intervals = siemens_clean["timestamp"].diff()

# Find intervals longer than one second.
long_gap_indices = timestamp_intervals[
    timestamp_intervals > 1000
].index

print("Number of gaps longer than 1 second:", len(long_gap_indices))

for index in long_gap_indices:
    previous_timestamp = siemens_clean.loc[index - 1, "timestamp"]
    current_timestamp = siemens_clean.loc[index, "timestamp"]
    gap_ms = timestamp_intervals.loc[index]

    print(f"\nPrevious timestamp: {previous_timestamp}")
    print(f"Current timestamp:  {current_timestamp}")
    print(f"Gap: {gap_ms} ms")
```

```python
# Final basic data-quality check.
print("Dataset shape:", siemens_clean.shape)

print(
    "Total missing values:",
    siemens_clean.isna().sum().sum()
)

print(
    "Timestamps strictly increasing:",
    siemens_clean["timestamp"].is_monotonic_increasing
)

print(
    "Duplicate timestamps:",
    siemens_clean["timestamp"].duplicated().sum()
)
```

**Validation results**
- No missing values remain in the working dataset.
- Timestamps are strictly increasing.
- No duplicate timestamps were found.

**Timestamp observations**

**Remaining preprocessing decision**
